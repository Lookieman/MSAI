```python
# train.py
from torch.cuda import amp
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
import torch.nn.functional as F
from tqdm import tqdm
import numpy as np
import os
from torch.utils.tensorboard import SummaryWriter  # For TensorBoard
import copy  # For Early Stopping
import matplotlib.pyplot as plt
import torchvision  # For visualization
from PIL import Image  # Add this import
import multiprocessing
import gc


# =============================
# Configuration Dictionary
# =============================

config = {
    "use_dropout": False,
    "dropout_rate": 0.3,
    "use_weight_decay": True,
    "weight_decay": 5e-4,
    "use_data_augmentation": True,
    "use_stochastic_depth": False,
    "stochastic_depth_prob": 0.2,
    "use_label_smoothing": False,  # Set to False if using Mixup
    "label_smoothing": 0.1,
    "use_mixup": False,  # Set to False if using Label Smoothing
    "mixup_alpha": 0.4,
    "use_spt": False,  # Shifted Patch Tokenization
    "use_lsa": False,  # Locality Self-Attention
    "use_augmented_conv": False,
    "lsa_window_size": 5,
    "dk": 64,  # dimension of key (add this)
    "dv": 64,  # dimension of value (add this)
    "learning_rate": 0.1, # Base learning rate before scaling
    "epochs": 100,
    "batch_size": 128,
    "dataset_mode": "CIFAR100",  # Choose between CIFAR10 or CIFAR100
    "warmup_epochs": 25,  # 5% of 500 epochs
    "warmup_start_lr": 0.0,  # Starting learning rate for warmup
    "lr_scheduler": "warmup_cosine",  # Options: 'step', 'cosine', 'reduce_on_plateau', "warmip_cosine", 'none'
    "eta_min": 0.00001,        # Minimum LR
    "step_size": 50,
    "gamma": 0.1,
    "checkpoint_dir": "checkpoints",  # Directory to save checkpoints
    "resume_checkpoint": "None",  # Path to resume checkpoint (set manually)
    "early_stopping_patience": float('inf'),  # Number of epochs to wait for improvement
    #cutout parameters
    "cutout": {
        "enabled": True,
        "n_holes": 1,
        "length": 16
    },
    #SPT parameters
    "shift": {
        "shifts": (1, 1),
        "dims": (1, 2)
    }
}

# Ensure checkpoint directory exists
os.makedirs(config["checkpoint_dir"], exist_ok=True)

# Initialize TensorBoard SummaryWriter
writer = SummaryWriter(log_dir='runs/CIFAR100_experiment')

# =============================
# Device Configuration
# =============================

device = torch.device("cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu")
print(f"Using device: {device}")
print(f"LSA enabled: {config['use_lsa']}")
print(f"LSA window size: {config['lsa_window_size']}")

# =============================
# Unpack Configuration Values
# =============================

dropout_rate = config["dropout_rate"] if config["use_dropout"] else 0.0
weight_decay = config["weight_decay"] if config["use_weight_decay"] else 0.0
mixup_alpha = config["mixup_alpha"] if config["use_mixup"] else 0.0
label_smoothing = config["label_smoothing"] if config["use_label_smoothing"] else 0.0
use_spt = config["use_spt"]
use_lsa = config["use_lsa"]
use_augmented_conv = config["use_augmented_conv"]
lsa_window_size = config["lsa_window_size"]
cutout_config = config.get("cutout", {})
use_cutout = cutout_config.get("enabled", False)
cutout_n_holes = cutout_config.get("n_holes", 1)
cutout_length = cutout_config.get("length", 16)
lsa_window_size = config["lsa_window_size"]


# =============================
# Top-Level Function for Shifting
# =============================

class Shift(object):
    """Shift the tensor by specified shifts in given dimensions."""
    def __init__(self, shifts=(1, 1), dims=(1, 2)):
        """
        Args:
            shifts (tuple): The number of places to shift in each dimension.
            dims (tuple): The dimensions along which to shift.
        """
        self.shifts = shifts
        self.dims = dims

    def __call__(self, img):
        if isinstance(img, Image.Image):
            # Convert PIL Image to tensor
            to_tensor = transforms.ToTensor()
            img = to_tensor(img)
        return torch.roll(img, shifts=self.shifts, dims=self.dims)

# =============================
# Cutout Data Augmentation
# =============================

class Cutout(object):
    """Randomly mask out one or more square regions of an image."""
    def __init__(self, n_holes=1, length=16):
        self.n_holes = n_holes
        self.length = length

    def __call__(self, img):
        """
        Args:
            img: PIL Image or Tensor
        Returns:
            PIL Image or Tensor: Image with n_holes of dimension length x length cut out
        """
        if isinstance(img, Image.Image):
            # Convert PIL Image to tensor
            to_tensor = transforms.ToTensor()
            img = to_tensor(img)

        # Handle tensor input
        if torch.is_tensor(img):
            h = img.size(1)
            w = img.size(2)

            mask = torch.ones((h, w), device=img.device)

            for _ in range(self.n_holes):
                y = torch.randint(h, (1,)).item()
                x = torch.randint(w, (1,)).item()

                y1 = max(y - self.length // 2, 0)
                y2 = min(y + self.length // 2, h)
                x1 = max(x - self.length // 2, 0)
                x2 = min(x + self.length // 2, w)

                mask[y1:y2, x1:x2] = 0

            mask = mask.expand_as(img)
            img = img * mask

        return img

    def __getstate__(self):
        return {'n_holes': self.n_holes, 'length': self.length}

    def __setstate__(self, state):
        self.n_holes = state['n_holes']
        self.length = state['length']


# =============================
# Model Definitions
# =============================

class BasicBlock(nn.Module):
    def __init__(self, in_planes, out_planes, stride, dropout_rate=0.0, use_augmented_conv=False, config=None):
        super(BasicBlock, self).__init__()
        self.use_augmented_conv = use_augmented_conv

        # Use AugmentedConv for the first layer if enabled
        if self.use_augmented_conv:
            self.conv1 = AugmentedConv(
                in_channels=in_planes,
                out_channels=out_planes,
                kernel_size=3,
                dk=16,
                dv=16,
                Nh=4,  # number of heads
                shape=32,  # assuming input shape is 32x32
                relative=True,
                stride=stride,
                use_lsa=config["use_lsa"],
                window_size=config["lsa_window_size"]
            )
        else:
            self.conv1 = nn.Conv2d(in_planes, out_planes, kernel_size=3, stride=stride, padding=1, bias=False)

        self.bn1 = nn.BatchNorm2d(in_planes)
        self.dropout = nn.Dropout(dropout_rate)

        # Second layer, use standard convolution
        self.bn2 = nn.BatchNorm2d(out_planes)
        self.conv2 = nn.Conv2d(out_planes, out_planes, kernel_size=3, stride=1, padding=1, bias=False)

        # Shortcut layer
        self.shortcut = nn.Sequential()
        if stride != 1 or in_planes != out_planes:
            self.shortcut = nn.Sequential(
                nn.Conv2d(in_planes, out_planes, kernel_size=1, stride=stride, bias=False),
            )

    def forward(self, x):
        out = F.relu(self.bn1(x))
        out = self.conv1(out) if self.use_augmented_conv else self.conv1(F.relu(self.bn1(x)))
        out = self.dropout(out)
        out = self.conv2(F.relu(self.bn2(out)))
        out += self.shortcut(x)
        return out

class Wide_ResNet(nn.Module):
    def __init__(self, depth, widen_factor, dropout_rate, num_classes, shape=32,
                 use_augmented_conv=False, use_lsa=False, lsa_window_size=3, config=None):
        super(Wide_ResNet, self).__init__()
        self.in_planes = 16
        self.use_augmented_conv = use_augmented_conv
        self.use_lsa = use_lsa
        self.lsa_window_size = lsa_window_size

        n = (depth - 4) // 6
        k = widen_factor

        self.conv1 = nn.Conv2d(3, 16, kernel_size=3, stride=1, padding=1, bias=False)
        self.layer1 = self._make_layer(16 * k, n, dropout_rate, stride=1, use_augmented_conv=use_augmented_conv, config=config)
        self.layer2 = self._make_layer(32 * k, n, dropout_rate, stride=2, use_augmented_conv=use_augmented_conv, config=config)
        self.layer3 = self._make_layer(64 * k, n, dropout_rate, stride=2, use_augmented_conv=use_augmented_conv, config=config)
        self.bn = nn.BatchNorm2d(64 * k)
        self.fc = nn.Linear(64 * k, num_classes)

    def _make_layer(self, out_planes, blocks, dropout_rate, stride, use_augmented_conv=False, config=None):
        strides = [stride] + [1] * (blocks - 1)
        layers = []
        for stride in strides:
            layers.append(BasicBlock(self.in_planes, out_planes, stride, dropout_rate, use_augmented_conv=use_augmented_conv, config=config))
            self.in_planes = out_planes
        return nn.Sequential(*layers)

    def forward(self, x):
        out = self.conv1(x)
        out = self.layer1(out)
        out = self.layer2(out)
        out = self.layer3(out)
        out = F.relu(self.bn(out))
        out = F.avg_pool2d(out, out.size(3))
        out = out.view(out.size(0), -1)
        out = self.fc(out)
        return out

# =============================
# AugmentedConv Layer with LSA
# =============================

class AugmentedConv(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size, dk, dv, Nh, shape=0, relative=False, stride=1, use_lsa=False, window_size=3):
        super(AugmentedConv, self).__init__()
        self.in_channels = in_channels
        self.out_channels = out_channels
        self.kernel_size = kernel_size
        self.dk = dk
        self.dv = dv
        self.Nh = Nh
        self.shape = shape
        self.relative = relative
        self.stride = stride
        self.use_lsa = use_lsa
        self.window_size = window_size
        self.padding = (self.kernel_size - 1) // 2

        # Convolution layers
        self.conv_out = nn.Conv2d(self.in_channels, self.out_channels - self.dv, self.kernel_size, stride=stride, padding=self.padding)
        self.qkv_conv = nn.Conv2d(self.in_channels, 2 * self.dk + self.dv, kernel_size=self.kernel_size, stride=stride, padding=self.padding)
        self.attn_out = nn.Conv2d(self.dv, self.dv, kernel_size=1, stride=1)

        # Relative positional encoding parameters
        if self.relative:
            self.key_rel_w = nn.Parameter(torch.randn((2 * self.shape - 1, dk // Nh), requires_grad=True))
            self.key_rel_h = nn.Parameter(torch.randn((2 * self.shape - 1, dk // Nh), requires_grad=True))

    def forward(self, x):
        conv_out = self.conv_out(x)
        N, _, H, W = conv_out.size()

        # Compute q, k, v
        qkv = self.qkv_conv(x)
        q, k, v = torch.split(qkv, [self.dk, self.dk, self.dv], dim=1)

        # Split heads
        q = self.split_heads_2d(q, self.Nh)  # [N, Nh, dk//Nh, H, W]
        k = self.split_heads_2d(k, self.Nh)  # [N, Nh, dk//Nh, H, W]
        v = self.split_heads_2d(v, self.Nh)  # [N, Nh, dv//Nh, H, W]

        if self.use_lsa:
            attn_out = self.compute_local_attention(q, k, v)
        else:
            # Global attention
            flat_q = q.view(N, self.Nh, self.dk // self.Nh, -1)
            flat_k = k.view(N, self.Nh, self.dk // self.Nh, -1)
            flat_v = v.view(N, self.Nh, self.dv // self.Nh, -1)

            attn = torch.matmul(flat_q.transpose(-2, -1), flat_k)
            attn = F.softmax(attn / (self.dk ** 0.5), dim=-1)
            attn_out = torch.matmul(attn, flat_v.transpose(-2, -1))
            attn_out = attn_out.view(N, self.Nh, self.dv // self.Nh, H, W)

        # Combine heads and process through final convolution
        attn_out = self.combine_heads_2d(attn_out)
        attn_out = self.attn_out(attn_out)

        # Concatenate with convolution output
        return torch.cat((conv_out, attn_out), dim=1)

    def compute_flat_qkv(self, x, dk, dv, Nh):
        qkv = self.qkv_conv(x)
        N, _, H, W = qkv.size()
        q, k, v = torch.split(qkv, [dk, dk, dv], dim=1)

        # Reshape q, k, v to have same dimensions
        q = self.split_heads_2d(q, Nh)  # (N, Nh, dk//Nh, H, W)
        k = self.split_heads_2d(k, Nh)  # (N, Nh, dk//Nh, H, W)
        v = self.split_heads_2d(v, Nh)  # (N, Nh, dv//Nh, H, W)

        dk_per_head = dk // Nh
        q = q * (dk_per_head ** -0.5)  # Scale q

        # Reshape for attention computation
        flat_q = q.reshape(N, Nh, dk_per_head, H * W)
        flat_k = k.reshape(N, Nh, dk_per_head, H * W)
        flat_v = v.reshape(N, Nh, dv // Nh, H * W)

        return flat_q, flat_k, flat_v, q, k, v

    def split_heads_2d(self, x, Nh):
        return x.view(x.size(0), Nh, x.size(1) // Nh, *x.shape[2:])

    def combine_heads_2d(self, x):
        return x.view(x.size(0), -1, *x.shape[3:])

    def compute_local_attention(self, q, k, v):
        N, Nh, dkh, H, W = q.size()

        # Process in chunks to save memory
        chunk_size = min(H, 8)
        attn_out = torch.zeros(N, Nh, self.dv // self.Nh, H, W, device=q.device)

        for i in range(0, H, chunk_size):
            i_end = min(i + chunk_size, H)
            for j in range(W):
                # Window boundaries
                i_min = max(0, i - self.window_size // 2)
                i_max = min(H, i_end + self.window_size // 2)
                j_min = max(0, j - self.window_size // 2)
                j_max = min(W, j + self.window_size // 2)

                # Extract local regions
                q_local = q[:, :, :, i:i_end, j].reshape(N, Nh, dkh, -1)  # [N, Nh, dkh, chunk_size]
                k_local = k[:, :, :, i_min:i_max, j_min:j_max].reshape(N, Nh, dkh, -1)  # [N, Nh, dkh, window_size*window_size]
                v_local = v[:, :, :, i_min:i_max, j_min:j_max].reshape(N, Nh, self.dv // self.Nh, -1)  # [N, Nh, dv//Nh, window_size*window_size]

                # Compute attention scores - proper transposition for matrix multiplication
                attn = torch.matmul(q_local.transpose(2, 3), k_local) / np.sqrt(dkh)  # [N, Nh, chunk_size, window_size*window_size]
                attn = F.softmax(attn, dim=-1)

                # Apply attention to values
                out = torch.matmul(attn, v_local.transpose(2, 3))  # [N, Nh, chunk_size, dv//Nh]

                # Reshape back
                attn_out[:, :, :, i:i_end, j] = out.transpose(2, 3)  # Move channels back to correct dimension

        return attn_out

    def relative_logits(self, q):
        # Compute relative logits for height and width
        B, Nh, dkh, H, W = q.shape
        rel_logits_w = self.relative_logits_1d(q, self.key_rel_w, H, W, transpose_mask=(0, 1, 2, 4, 3, 5))
        rel_logits_h = self.relative_logits_1d(q.permute(0, 1, 2, 4, 3), self.key_rel_h, W, H, transpose_mask=(0, 1, 2, 4, 5, 3))
        return rel_logits_w, rel_logits_h

    def relative_logits_1d(self, q, rel_k, H, W, transpose_mask):
        # Compute 1D relative logits along a single axis (height or width)
        rel_logits = torch.einsum('bhxyd,md->bhxym', q, rel_k)
        rel_logits = rel_logits.view(-1, self.Nh, H, W, 2 * H - 1)
        rel_logits = self._relative_to_absolute(rel_logits)
        rel_logits = rel_logits.view(-1, self.Nh, H, W, W).permute(transpose_mask)
        return rel_logits

    def _relative_to_absolute(self, x):
        # Convert relative indices to absolute indices
        B, Nh, H, L, _ = x.size()
        col_pad = torch.zeros((B, Nh, H, L, 1), device=x.device, dtype=x.dtype)
        x = torch.cat([x, col_pad], dim=-1)
        flat_x = x.view(B, Nh, H, -1)
        flat_pad = torch.zeros((B, Nh, H, L - 1), device=x.device, dtype=x.dtype)
        flat_x = torch.cat([flat_x, flat_pad], dim=-1)
        final_x = flat_x.view(B, Nh, H, L + 1, 2 * L - 1)[:, :, :, :L, L - 1:]
        return final_x

# =============================
# Learning Rate Scheduler with Warmup
# =============================

class WarmupCosineLR:
    """Custom learning rate scheduler with warmup and cosine decay."""
    def __init__(self, optimizer, warmup_epochs, total_epochs, base_lr):
        self.optimizer = optimizer
        self.warmup_epochs = warmup_epochs
        self.total_epochs = total_epochs
        self.base_lr = base_lr

        # Calculate target lr with batch size scaling
        self.target_lr = self.base_lr

        # Create cosine scheduler for after warmup
        self.cosine_scheduler = optim.lr_scheduler.CosineAnnealingLR(
            optimizer,
            T_max=total_epochs - warmup_epochs,
            eta_min=config["eta_min"]
        )

        self.current_lr = config["warmup_start_lr"]
        self._update_optimizer()

    def _update_optimizer(self):
        """Update the learning rate in the optimizer."""
        for param_group in self.optimizer.param_groups:
            param_group['lr'] = self.current_lr

    def step(self, epoch):
        """Update learning rate based on current epoch."""
        if epoch < self.warmup_epochs:
            # Linear warmup
            self.current_lr = self.target_lr * (epoch / self.warmup_epochs)
            self._update_optimizer()
        else:
            # Cosine decay
            self.cosine_scheduler.step()
            self.current_lr = self.optimizer.param_groups[0]['lr']

    def get_lr(self):
        """Get current learning rate."""
        return self.current_lr

    def state_dict(self):
        """Returns the state of the scheduler as a :class:`dict`.

        It contains an entry for every variable in self.__dict__ which
        is not the optimizer.
        """
        return {
            'warmup_epochs': self.warmup_epochs,
            'total_epochs': self.total_epochs,
            'base_lr': self.base_lr,
            'target_lr': self.target_lr,
            'current_lr': self.current_lr,
            'cosine_scheduler': self.cosine_scheduler.state_dict()
        }

    def load_state_dict(self, state_dict):
        """Loads the schedulers state.

        Arguments:
            state_dict (dict): scheduler state. Should be an object returned
                from a call to :meth:`state_dict`.
        """
        self.warmup_epochs = state_dict['warmup_epochs']
        self.total_epochs = state_dict['total_epochs']
        self.base_lr = state_dict['base_lr']
        self.target_lr = state_dict['target_lr']
        self.current_lr = state_dict['current_lr']
        self.cosine_scheduler.load_state_dict(state_dict['cosine_scheduler'])


# =============================
# Utility Functions
# =============================

def load_data(dataset_mode, batch_size=128, data_augmentation=False, use_spt=False, val_split=0.1):
    """
    Load CIFAR10 or CIFAR100 dataset with train/val/test splits.
    val_split: proportion of training data to use for validation
    """
    transform_train = [
        transforms.RandomCrop(32, padding=4),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor()
    ]

    if data_augmentation:
        transform_train.insert(-1, transforms.AutoAugment(policy=transforms.AutoAugmentPolicy.CIFAR10))

    if use_cutout:
        transform_train.append(Cutout(n_holes=cutout_n_holes, length=cutout_length))

    if use_spt:
        transform_train.append(Shift(
            shifts=config["shift"]["shifts"],
            dims=config["shift"]["dims"]
        ))

    transform_train.append(
        transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010))
    )

    transform_train = transforms.Compose(transform_train)

    transform_eval = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010))
    ])

    # Load the appropriate dataset
    if dataset_mode == "CIFAR10":
        train_full = datasets.CIFAR10('data', train=True, download=True, transform=transform_train)
        test_dataset = datasets.CIFAR10('data', train=False, download=True, transform=transform_eval)
    elif dataset_mode == "CIFAR100":
        train_full = datasets.CIFAR100('data', train=True, download=True, transform=transform_train)
        test_dataset = datasets.CIFAR100('data', train=False, download=True, transform=transform_eval)
    else:
        raise ValueError("Unsupported dataset mode")

    # Calculate split sizes
    total_size = len(train_full)
    val_size = int(total_size * val_split)
    train_size = total_size - val_size

    # Split training dataset into train and validation sets
    train_dataset, val_dataset = torch.utils.data.random_split(
        train_full,
        [train_size, val_size],
        generator=torch.Generator().manual_seed(42)  # For reproducibility
    )

    # Create data loaders
    train_loader = torch.utils.data.DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        num_workers=0,  # Set to 0 for debugging
        pin_memory=True
    )

    val_loader = torch.utils.data.DataLoader(
        val_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=0,
        pin_memory=True
    )

    test_loader = torch.utils.data.DataLoader(
        test_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=0,
        pin_memory=True
    )

    return train_loader, val_loader, test_loader

def mixup_data(x, y, alpha=0.4):
    """Apply Mixup augmentation."""
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1
    batch_size = x.size()[0]
    index = torch.randperm(batch_size).to(x.device)
    mixed_x = lam * x + (1 - lam) * x[index, :]
    y_a, y_b = y, y[index]
    return mixed_x, y_a, y_b, lam

def smooth_labels(labels, num_classes, smoothing=0.1):
    """Apply label smoothing."""
    confidence = 1.0 - smoothing
    label_shape = torch.Size((labels.size(0), num_classes))
    with torch.no_grad():
        true_dist = torch.full(label_shape, smoothing / (num_classes - 1)).to(labels.device)
        true_dist.scatter_(1, labels.data.unsqueeze(1), confidence)
    return true_dist

def calculate_accuracy(output, target, top_k=(1, 5)):
    """Calculate Top-K accuracy."""
    with torch.no_grad():
        max_k = max(top_k)
        batch_size = target.size(0)

        # Ensure target is on the same device as output
        target = target.to(output.device)

        _, pred = output.topk(max_k, 1, True, True)
        pred = pred.t()
        correct = pred.eq(target.view(1, -1).expand_as(pred))

        res = []
        for k in top_k:
            correct_k = correct[:k].reshape(-1).float().sum(0, keepdim=True)
            res.append(correct_k.mul_(100.0 / batch_size))
        return res  # Returns [top1_acc, top5_acc]

def save_checkpoint(state, filename):
    """Save the training checkpoint."""
    torch.save(state, filename)
    print(f"Checkpoint saved to {filename}")

def load_checkpoint(model, optimizer, scheduler, filename):
    """Load the training checkpoint."""
    if not os.path.isfile(filename):
        print(f"No checkpoint found at '{filename}'")
        return 0, 0.0  # Start from epoch 0 with 0 best accuracy
    checkpoint = torch.load(filename, map_location=device)
    model.load_state_dict(checkpoint['model_state_dict'])
    optimizer.load_state_dict(checkpoint['optimizer_state_dict'])
    if scheduler and 'scheduler_state_dict' in checkpoint and checkpoint['scheduler_state_dict'] is not None:
        scheduler.load_state_dict(checkpoint['scheduler_state_dict'])
    start_epoch = checkpoint['epoch'] + 1
    best_val_acc = checkpoint.get('best_val_acc', 0.0)
    print(f"Checkpoint loaded. Resuming from epoch {start_epoch}. Best Validation Accuracy: {best_val_acc:.2f}%")
    return start_epoch, best_val_acc



# =============================
# Early Stopping Class
# =============================

class EarlyStopping:
    """Early stops the training if validation accuracy doesn't improve after a given patience."""
    def __init__(self, patience=20, verbose=False, delta=0.0, path='best_model.pth'):
        """
        Args:
            patience (int): How long to wait after last time validation accuracy improved.
            verbose (bool): If True, prints a message for each validation accuracy improvement.
            delta (float): Minimum change in the monitored quantity to qualify as an improvement.
            path (str): Path for the checkpoint to be saved to.
        """
        self.patience = patience
        self.verbose = verbose
        self.counter = 0
        self.best_score = None
        self.early_stop = False
        self.val_acc_max = 0.0
        self.delta = delta
        self.path = path

    def __call__(self, val_acc, model):

        score = val_acc

        if self.best_score is None:
            self.best_score = score
            self.save_checkpoint(val_acc, model)
        elif score < self.best_score + self.delta:
            self.counter += 1
            if self.verbose:
                print(f'EarlyStopping counter: {self.counter} out of {self.patience}')
            if self.counter >= self.patience:
                self.early_stop = True
        else:
            self.best_score = score
            self.save_checkpoint(val_acc, model)
            self.counter = 0

    def save_checkpoint(self, val_acc, model):
        """Saves model when validation accuracy increases."""
        if self.verbose:
            print(f'Validation accuracy increased ({self.val_acc_max:.6f} --> {val_acc:.6f}).  Saving model ...')
        torch.save(model.state_dict(), self.path)
        self.val_acc_max = val_acc

# =============================
# Training Function
# =============================

def train(model, train_loader, optimizer, criterion, epoch, num_classes):

    model.train()
    torch.cuda.empty_cache()  # Add here
    gc.collect()
    total_loss = 0
    total_top1_acc = 0
    total_top5_acc = 0  # Add top-5 tracking

    # Add device type check for amp
    scaler = None
    if device.type == 'cuda':
        scaler = amp.GradScaler()
    elif device.type == 'mps':
        # MPS doesn't support AMP yet, so we'll skip it
        pass
    else:
        # CPU doesn't need AMP
        pass

    progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}", unit="batch", leave=False)

    for batch_idx, (data, target) in enumerate(progress_bar):
        # Add shape debugging at first batch of first epoch
        if epoch == 0 and batch_idx == 0:
            print(f"Shape before LSA: {data.shape}")

        # Apply mixup if enabled
        if config["use_mixup"]:
            data, targets_a, targets_b, lam = mixup_data(data, target, mixup_alpha)
            data, targets_a, targets_b = data.to(device), targets_a.to(device), targets_b.to(device)

            # Conditional AMP usage
            if scaler is not None:
                with amp.autocast():
                    output = model(data)
                    if epoch == 0 and batch_idx == 0:
                        print(f"Shape after LSA: {output.shape}")
                    loss = lam * criterion(output, targets_a) + (1 - lam) * criterion(output, targets_b)
            else:
                output = model(data)
                if epoch == 0 and batch_idx == 0:
                    print(f"Shape after LSA: {output.shape}")
                loss = lam * criterion(output, targets_a) + (1 - lam) * criterion(output, targets_b)

            # Calculate both top-1 and top-5 accuracies
            accs_a = calculate_accuracy(output, targets_a)
            accs_b = calculate_accuracy(output, targets_b)
            top1_acc = lam * accs_a[0] + (1 - lam) * accs_b[0]
            top5_acc = lam * accs_a[1] + (1 - lam) * accs_b[1]
        else:
            data, target = data.to(device), target.to(device)

            # Conditional AMP usage
            if scaler is not None:
                with amp.autocast():
                    output = model(data)
                    if epoch == 0 and batch_idx == 0:
                        print(f"Shape after LSA: {output.shape}")
                    loss = criterion(output, target)
            else:
                output = model(data)
                if epoch == 0 and batch_idx == 0:
                    print(f"Shape after LSA: {output.shape}")
                loss = criterion(output, target)


            # Calculate both top-1 and top-5 accuracies
            accs = calculate_accuracy(output, target)
            top1_acc = accs[0]
            top5_acc = accs[1]

        # Modified optimization step with conditional gradient scaling
        optimizer.zero_grad()

        if scaler is not None:
            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
            scaler.step(optimizer)
            scaler.update()
        else:
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
            optimizer.step()

        # Add cleanup here
        if batch_idx % 100 == 0:  # Every 100 batches
            torch.cuda.empty_cache()
            gc.collect()

        total_loss += loss.item()
        total_top1_acc += top1_acc.item()
        total_top5_acc += top5_acc.item()  # Track top-5 accuracy

        progress_bar.set_postfix(
            loss=loss.item(),
            top1_acc=top1_acc.item(),
            top5_acc=top5_acc.item()
        )

    avg_loss = total_loss / len(train_loader)
    avg_top1_acc = total_top1_acc / len(train_loader)
    avg_top5_acc = total_top5_acc / len(train_loader)

    print(f"Epoch {epoch + 1} - Training Loss: {avg_loss:.4f}, "
          f"Top-1 Accuracy: {avg_top1_acc:.2f}%, "
          f"Top-5 Accuracy: {avg_top5_acc:.2f}%")

    # Log to TensorBoard
    writer.add_scalar('Loss/Train', avg_loss, epoch)
    writer.add_scalar('Accuracy/Train_Top1', avg_top1_acc, epoch)
    writer.add_scalar('Accuracy/Train_Top5', avg_top5_acc, epoch)

    return avg_loss, avg_top1_acc, avg_top5_acc


# =============================
# Evaluation Function
# =============================

def evaluate(model, test_loader, num_classes, epoch):
    """Evaluation loop for validation."""
    model.eval()
    total_top1_acc = 0  # Track total Top-1 accuracy for the validation set
    total_top5_acc = 0  # Add top-5 tracking
    total_loss = 0  # Track total loss for the validation set
    criterion = nn.CrossEntropyLoss(label_smoothing=label_smoothing).to(device) if config["use_label_smoothing"] else nn.CrossEntropyLoss().to(device)

    with torch.no_grad():
        for data, target in test_loader:
            data, target = data.to(device), target.to(device)
            output = model(data)
            loss = criterion(output, target)
            total_loss += loss.item()

            # Calculate both top-1 and top-5 accuracies
            accs = calculate_accuracy(output, target)
            total_top1_acc += accs[0].item()
            total_top5_acc += accs[1].item()

    avg_loss = total_loss / len(test_loader)
    avg_top1_acc = total_top1_acc / len(test_loader)
    avg_top5_acc = total_top5_acc / len(test_loader)

    print(f"Validation Loss: {avg_loss:.4f}, "
          f"Top-1 Accuracy: {avg_top1_acc:.2f}%, "
          f"Top-5 Accuracy: {avg_top5_acc:.2f}%")

    # Log to TensorBoard
    writer.add_scalar('Loss/Validation', avg_loss, epoch)
    writer.add_scalar('Accuracy/Val_Top1', avg_top1_acc, epoch)
    writer.add_scalar('Accuracy/Val_Top5', avg_top5_acc, epoch)

    return avg_loss, avg_top1_acc, avg_top5_acc

# =============================
# Visualization Function
# =============================

def visualize_augmented_images(loader, num_images=5):
    """Visualize augmented images"""
    try:
        # Get a batch of images
        images, _ = next(iter(loader))
        images = images[:num_images]

        # Create grid
        grid = torchvision.utils.make_grid(
            images,
            nrow=num_images,
            padding=2,
            normalize=True,
            value_range=(-2, 2)
        )

        # Convert to numpy for plotting
        np_grid = grid.cpu().numpy()
        np_grid = np.transpose(np_grid, (1, 2, 0))

        # Plot
        plt.figure(figsize=(15, 3))
        plt.imshow(np.clip(np_grid, 0, 1))
        plt.title("Sample Augmented Images")
        plt.axis('off')
        plt.show()
        plt.close()

    except Exception as e:
        print(f"Error in visualization: {str(e)}")
        import traceback
        traceback.print_exc()


# =============================
# Main Function
# =============================

def main():

    # Set multiprocessing start method at the beginning
    try:
        multiprocessing.set_start_method('spawn', force=True)
    except RuntimeError:
        # Ignore if already set
        pass

    # After device setup
    if device.type == 'cuda':
        torch.cuda.empty_cache()
        gc.collect()

    print(f"Using device: {device}")
    print(f"Training with augmented configuration: {config['use_data_augmentation']}")
    print(f"Batch Size: {config['batch_size']}")
    print(f"Weight Decay: {config['weight_decay']}")
    print(f"Base Learning Rate: {config['learning_rate']}")
    print(f"Total Epochs: {config['epochs']}")
    print(f"Warmup Epochs: {config['warmup_epochs']}")

    # Learning rate initialization based on scheduler type
    initial_lr = (config["warmup_start_lr"]
                 if config["lr_scheduler"] == "warmup_cosine"
                 else config["learning_rate"])

    print("\nLearning Rate Configuration:")
    print(f"Scheduler Type: {config['lr_scheduler']}")
    print(f"Initial Learning Rate: {initial_lr}")
    if config["lr_scheduler"] == "warmup_cosine":
        print(f"Warmup Start LR: {config['warmup_start_lr']}")
        print(f"Warmup Epochs: {config['warmup_epochs']}")
        print(f"Target LR after warmup: {config['learning_rate']}")
    print(f"Total Epochs: {config['epochs']}")


    # Load data with three-way split
    train_loader, val_loader, test_loader = load_data(
        dataset_mode=config["dataset_mode"],
        batch_size=config["batch_size"],
        data_augmentation=config["use_data_augmentation"],
        use_spt=config["use_spt"]
    )

    # Model initialization
    num_classes = 10 if config["dataset_mode"] == "CIFAR10" else 100
    model = Wide_ResNet(
        depth=28,
        widen_factor=10,
        dropout_rate=dropout_rate,
        num_classes=num_classes,
        shape=32,
        use_augmented_conv=config["use_augmented_conv"],
        use_lsa=config["use_lsa"],
        lsa_window_size=config["lsa_window_size"],
        config=config
    ).to(device)

    # Try visualization with proper error handling
    try:
        print("Visualizing augmented images...")
        visualize_augmented_images(train_loader)
    except Exception as e:
        print(f"Visualization error (non-critical): {str(e)}")
        print("Continuing with training...")

    # Optimizer definition
    optimizer = optim.SGD(
        model.parameters(),
        lr=config["learning_rate"],
        momentum=0.9,
        weight_decay=weight_decay,
        nesterov=True
    )

    # Scheduler definition
    scheduler = None
    if config["lr_scheduler"] == "step":
        scheduler = optim.lr_scheduler.StepLR(
            optimizer,
            step_size=config["step_size"],
            gamma=config["gamma"]
        )
    elif config["lr_scheduler"] == "cosine":
        scheduler = optim.lr_scheduler.CosineAnnealingLR(
            optimizer,
            T_max=config["epochs"],
            eta_min=config.get("eta_min", 0.0)
        )
    elif config["lr_scheduler"] == "reduce_on_plateau":
        scheduler = optim.lr_scheduler.ReduceLROnPlateau(
            optimizer,
            mode='min',
            factor=0.1,
            patience=10,
            verbose=True
        )
    elif config["lr_scheduler"] == "warmup_cosine":
        scheduler = WarmupCosineLR(
            optimizer,
            warmup_epochs=config["warmup_epochs"],
            total_epochs=config["epochs"],
            base_lr=config["learning_rate"]
        )
    elif config["lr_scheduler"] == "none":
        scheduler = None
    else:
        raise ValueError(f"Unsupported lr_scheduler type: {config['lr_scheduler']}")

    # Criterion definition
    criterion = (
        nn.CrossEntropyLoss(label_smoothing=label_smoothing).to(device)
        if config["use_label_smoothing"]
        else nn.CrossEntropyLoss().to(device)
    )

    # Initialize training state
    start_epoch = 0
    best_val_acc = 0.0

    # Early Stopping initialization
    early_stopping = EarlyStopping(
        patience=config["early_stopping_patience"],
        verbose=True,
        delta=0.0,
        path=os.path.join(config["checkpoint_dir"], 'best_model.pth')
    )

    # Resume from checkpoint if specified
    if config["resume_checkpoint"]:
        start_epoch, best_val_acc = load_checkpoint(
            model,
            optimizer,
            scheduler,
            config["resume_checkpoint"]
        )
        # Adjust scheduler for remaining epochs if using cosine or warmup_cosine
        if config["lr_scheduler"] in ["cosine", "warmup_cosine"]:
            remaining_epochs = config["epochs"] - start_epoch
            if config["lr_scheduler"] == "cosine":
                scheduler = optim.lr_scheduler.CosineAnnealingLR(
                    optimizer,
                    T_max=remaining_epochs,
                    eta_min=config.get("eta_min", 0.0)
                )
            else:  # warmup_cosine
                scheduler = WarmupCosineLR(
                    optimizer=optimizer,
                    warmup_epochs=max(0, config["warmup_epochs"] - start_epoch),
                    total_epochs=remaining_epochs,
                    base_lr=config["learning_rate"]
                )

    # Training loop
    print(f"Starting training from epoch {start_epoch + 1}...")
    best_val_acc = 0.0
    best_val_top5_acc = 0.0
    test_top1_acc = 0.0
    test_top5_acc = 0.0

    try:
        for epoch in range(start_epoch, config["epochs"]):
            # Training phase
            train_loss, train_top1_acc, train_top5_acc = train(
                model,
                train_loader,
                optimizer,
                criterion,
                epoch,
                num_classes
            )

            # Validation phase
            val_loss, val_top1_acc, val_top5_acc = evaluate(
                model,
                val_loader,
                num_classes,
                epoch
            )

            # Log to TensorBoard
            writer.add_scalar('Loss/Train', train_loss, epoch)
            writer.add_scalar('Loss/Validation', val_loss, epoch)
            writer.add_scalar('Accuracy/Train', train_top1_acc, epoch)
            writer.add_scalar('Accuracy/Validation', val_top1_acc, epoch)

            # Scheduler step
            if scheduler:
                if config["lr_scheduler"] == "reduce_on_plateau":
                    scheduler.step(val_loss)
                elif config["lr_scheduler"] == "warmup_cosine":
                    scheduler.step(epoch)  # Pass epoch parameter
                else:
                    scheduler.step()

            # Log learning rate
            current_lr = optimizer.param_groups[0]['lr']
            writer.add_scalar('Learning_Rate', current_lr, epoch)
            print(f"Epoch {epoch + 1} - Current LR: {current_lr:.6f}")

            # Early Stopping check
            early_stopping(val_top1_acc, model)

            # Get current learning rate
            current_lr = optimizer.param_groups[0]['lr']

            # Save checkpoint for every epoch
            save_checkpoint({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'scheduler_state_dict': scheduler.state_dict() if scheduler else None,
                'training_stats': {
                    'epoch': epoch,
                    'train_loss': train_loss,
                    'train_top1_acc': train_top1_acc,
                    'train_top5_acc': train_top5_acc,
                    'val_loss': val_loss,
                    'val_top1_acc': val_top1_acc,
                    'val_top5_acc': val_top5_acc,
                    'learning_rate': current_lr
                }
            }, os.path.join(config["checkpoint_dir"], f'checkpoint_epoch_{epoch+1}.pth'))

            # Update best validation accuracy and save best model
            if val_top1_acc > best_val_acc:
                best_val_acc = val_top1_acc
                best_val_top5_acc = val_top5_acc  # Save corresponding top-5 acc
                save_checkpoint({
                    'epoch': epoch,
                    'model_state_dict': model.state_dict(),
                    'optimizer_state_dict': optimizer.state_dict(),
                    'scheduler_state_dict': scheduler.state_dict() if scheduler else None,
                    'best_val_acc': best_val_acc,
                    'best_val_top5_acc': best_val_top5_acc,
                    'train_top5_acc': train_top5_acc,
                    'training_stats': {
                        'epoch': epoch,
                        'train_loss': train_loss,
                        'train_top1_acc': train_top1_acc,
                        'train_top5_acc': train_top5_acc,
                        'val_loss': val_loss,
                        'val_top1_acc': val_top1_acc,
                        'val_top5_acc': val_top5_acc,
                        'learning_rate': current_lr
                    }
                }, os.path.join(config["checkpoint_dir"], 'best_model.pth'))

            if early_stopping.early_stop:
                print("Early stopping triggered. Stopping training.")
                break

        # After training completes, evaluate on test set
        print("Training completed. Evaluating best model on test set...")

        # Load best model
        best_model_path = os.path.join(config["checkpoint_dir"], 'best_model.pth')
        checkpoint = torch.load(best_model_path, map_location=device)
        model.load_state_dict(checkpoint['model_state_dict'])

        # Evaluate on test set
        test_loss, test_top1_acc, test_top5_acc = evaluate(model, test_loader, num_classes, epoch)
        print(f"Final Test Accuracies - Top-1: {test_top1_acc:.2f}%, Top-5: {test_top5_acc:.2f}%")

        # Log final metrics
        writer.add_hparams(
            {
                'learning_rate': config["learning_rate"],
                'batch_size': config["batch_size"],
                'dropout_rate': dropout_rate,
                'weight_decay': weight_decay
            },
            {
                'best_val_accuracy_top1': best_val_acc,
                'best_val_accuracy_top5': best_val_top5_acc,
                'final_test_accuracy_top1': test_top1_acc,
                'final_test_accuracy_top5': test_top5_acc
            }
        )

    except KeyboardInterrupt:
        print("\nTraining interrupted by user")
    except Exception as e:
        print(f"\nError during training: {str(e)}")
        raise
    finally:
        print(f"\nTraining completed.")
        print(f"Best Validation Accuracies - Top-1: {best_val_acc:.2f}%, Top-5: {best_val_top5_acc:.2f}%")
        print(f"Final Test Accuracies - Top-1: {test_top1_acc:.2f}%, Top-5: {test_top5_acc:.2f}%")

# =============================
# Run the Main Function
# =============================

if __name__ == "__main__":
    main()
```

```python
import os
import json
import pandas as pd
import torch
from glob import glob
import matplotlib.pyplot as plt

def extract_training_history():
    """
    Extract training history from checkpoint files and save to CSV/JSON.
    Returns a pandas DataFrame with the training history.
    """
    # Get all checkpoint files
    checkpoint_files = sorted(glob(os.path.join('checkpoints', 'checkpoint_epoch_*.pth')), 
                            key=lambda x: int(x.split('_')[-1].split('.')[0]))
    
    history = []
    
    for checkpoint_file in checkpoint_files:
        try:
            # Load checkpoint
            checkpoint = torch.load(checkpoint_file, map_location='cpu')
            
            # Extract training stats
            stats = checkpoint['training_stats']
            history.append({
                'epoch': stats['epoch'] + 1,  # Convert to 1-based indexing
                'train_loss': stats['train_loss'],
                'train_top1_acc': stats['train_top1_acc'],
                'train_top5_acc': stats['train_top5_acc'],
                'val_loss': stats['val_loss'],
                'val_top1_acc': stats['val_top1_acc'],
                'val_top5_acc': stats['val_top5_acc'],
                'learning_rate': stats['learning_rate']
            })
        except Exception as e:
            print(f"Error loading checkpoint {checkpoint_file}: {str(e)}")
    
    # Convert to DataFrame
    df = pd.DataFrame(history)
    
    # Save to CSV
    df.to_csv('training_history.csv', index=False)
    
    # Save to JSON with readable formatting
    with open('training_history.json', 'w') as f:
        json.dump(history, f, indent=4)
    
    print(f"Training history saved to training_history.csv and training_history.json")
    print("\nSummary of training:")
    print(f"Total epochs: {len(history)}")
    if len(history) > 0:
        print("\nTop-1 Accuracy:")
        print(f"Best validation accuracy: {max(df['val_top1_acc']):.2f}%")
        print(f"Final training accuracy: {df['train_top1_acc'].iloc[-1]:.2f}%")
        print(f"Final validation accuracy: {df['val_top1_acc'].iloc[-1]:.2f}%")
        
        print("\nTop-5 Accuracy:")
        print(f"Best validation accuracy: {max(df['val_top5_acc']):.2f}%")
        print(f"Final training accuracy: {df['train_top5_acc'].iloc[-1]:.2f}%")
        print(f"Final validation accuracy: {df['val_top5_acc'].iloc[-1]:.2f}%")
    
    return df

# Extract and plot the training history
if __name__ == "__main__":
    df = extract_training_history()
```

```python
def plot_training_history(df, save_individual=True):
    """
    Plot training metrics both individually and combined.
    Args:
        df: DataFrame with training history
        save_individual: Whether to save individual plots
    """
    os.makedirs('plots', exist_ok=True)
    plt.style.use('seaborn-v0_8')
    
    metrics = [
        ('top1_accuracy', 'Top-1 Accuracy over Epochs', 'Accuracy (%)', 
         [('train_top1_acc', 'Train'), ('val_top1_acc', 'Validation')]),
        ('top5_accuracy', 'Top-5 Accuracy over Epochs', 'Accuracy (%)', 
         [('train_top5_acc', 'Train'), ('val_top5_acc', 'Validation')]),
        ('loss', 'Loss over Epochs', 'Loss', 
         [('train_loss', 'Train'), ('val_loss', 'Validation')]),
        ('learning_rate', 'Learning Rate over Epochs', 'Learning Rate', 
         [('learning_rate', None)])
    ]
    
    # Create combined plot
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    axes = axes.ravel()
    
    for idx, (name, title, ylabel, data_cols) in enumerate(metrics):
        # Plot individual
        if save_individual:
            plt.figure(figsize=(10, 6))
            for col, label in data_cols:
                plt.plot(df['epoch'], df[col], label=label, marker='o')
            plt.title(title)
            plt.xlabel('Epoch')
            plt.ylabel(ylabel)
            if len(data_cols) > 1:
                plt.legend()
            plt.grid(True)
            plt.savefig(f'plots/{name}.png', dpi=300, bbox_inches='tight')
            plt.close()
        
        # Plot on combined figure
        for col, label in data_cols:
            axes[idx].plot(df['epoch'], df[col], label=label, marker='o')
        axes[idx].set_title(title)
        axes[idx].set_xlabel('Epoch')
        axes[idx].set_ylabel(ylabel)
        if len(data_cols) > 1:
            axes[idx].legend()
        axes[idx].grid(True)
    
    plt.tight_layout()
    plt.savefig('plots/training_history_combined.png', dpi=300, bbox_inches='tight')
    plt.show()

if __name__ == "__main__":
    plot_training_history(df)
```
